```python
def func(x):
    return 1 / (1 + np.exp(-x)) > 0.5
```

```python
class Perceptron:
    def __init__(self, epoch = 1000, learning_rate = 0.01):
        self.epoch = epoch
        self.lr = learning_rate
        self.weights = None

    def fit(self, x, y):
        x_ = np.insert(x, 0, 1, axis = 1)
        self.weights = np.ones(x_.shape[1])
        y_ = np.asarray(y)
        for i in range(self.epoch):
            indices = np.random.permutation(x_.shape[0])
            for index in indices:
                xi = x_[index]
                yi = y_[index]
                y_pred = np.dot(xi, self.weights)
                y_pred = func(y_pred)
                self.weights = self.weights  + self.lr * xi * (yi - y_pred)

        return self.weights

    def predict(self, x):
        x_ = np.insert(x, 0, 1, axis = 1)
        y_pred = np.dot(x_, self.weights)
        y_pred=  func(y_pred)
        return y_pred
```

```python
model = Perceptron()
model.fit(x_train, y_train)
```

```python
pred = model.predict(x_test)
accuracy_score(y_test, pred)
```

```python
class GD_perceptron:
    def __init__(self, lr = 0.01, epoch = 1000):
        self.epoch = epoch
        self.lr = lr
        self.weights = None

    def fit(self, x, y):
        x_ = np.insert(x, 0, 1, axis = 1)
        self.weights = np.ones(x_.shape[1])
        y_ = np.asarray(y)
        
        for j in range(self.epoch):
            for i in range(x.shape[0]):
                z = self.weights[0] * x_[i][0] + self.weights[1] * x_[i][1] + self.weights[2] * x_[i][2]
                if (z * y_[i]) <= 0:
                    self.weights[1] += self.lr * y_[i] * x_[i][1]
                    self.weights[2] += self.lr * y_[i] * x_[i][2]
                    self.weights[0] += self.lr * y_[i] * x_[i][0]
        return self.weights

    def predict(self, x):
        x_ = np.insert(x, 0, 1, axis = 1)
        y_pred = np.dot(x_, self.weights)
        y_pred = func(y_pred)
        return y_pred

        
```

```python
model = GD_perceptron()
model.fit(x_train, y_train)
```

```python
pred = model.predict(x_test)
accuracy_score(y_test, pred)
```
