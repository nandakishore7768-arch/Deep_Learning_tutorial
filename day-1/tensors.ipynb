creating a tensor

```python
#using empty to create a tensor
#this allocates memory in space and returns the value which is already present
# in this memory space

arr = torch.empty(2,3)
print(arr)
print(type(arr))
```

```python
#using zeros to create the tensor
#create a array of the given size filled with zeros

arr = torch.zeros(2,3)
print(arr)
print(type(arr))
```

```python
#using ones to create a tensor
# creates an array filled with ones

arr = torch.ones(2,3)
print(arr)
print(type(arr))
```

```python
#using torch.rand to create a random tensor of the given shape
#this value will not be repeated thus, we need to set the seed for this

arr = torch.rand(2,3)
print(arr)
print(type(arr))
```

```python
# setting the seed for the rand function

torch.manual_seed(100)
arr = torch.rand(2,3)
print(arr)
print(type(arr))
```

```python
#to create a custom tensor, we use torch.tensor

arr = torch.Tensor([[1,2,3],[4,5,6]])
print(arr)
print(type(arr))
```

```python
#other methods include torch.arange - creates an array from range provided

arr = torch.arange(0, 10)
print(arr)
```

```python
#torch.eye is used to get the identity matrix of a particular shape

arr = torch.eye(5)
print(arr)
```

```python
#using linspace to create an array
#divides the range equally into n divisions and returns the values

arr = torch.linspace(0, 10, 10)
print(arr.shape)
```

```python
# to create a fill matrix with a fill value, we use torch.full

arr = torch.full((3,3), 5)
print(arr)
```

type and shape of the tensor

```python
#getting the shape of the tensor

arr = torch.Tensor([[1,2,3], [4,5,6]])
arr.shape
```

```python
# creating an empty array of the same shape

arr1 = torch.empty(arr.shape)
arr1.shape
```

```python
#another method to create a array of same shape and size

arr1 = torch.empty_like(arr)
print(arr1)
print(arr1.shape)
```

```python
# creating a zero array of the same shape

arr1 = torch.zeros_like(arr)
print(arr1)
print(arr1.shape)
```

```python
#creating a random array of the same shape
#generates values between 0 and 1

arr1 = torch.rand_like(arr)
print(arr1, arr1.shape)
```

```python
# to find the data type of a tensor, we use dtype

arr.dtype
```

```python
arr1 = torch.tensor([[1,2,3] , [4,5,6]], dtype = torch.float32)
print(arr1.dtype)
arr_ = arr1.to(torch.int32)
print(arr_.dtype)

# there are two different constructors torch.tensor and torch.Tensor - the Tensor does not have dtype whereas tensor has a dtype argument
```
